# Getting started: Triton lower body on Antioch

Copy each Bash block into a **local terminal**, in order. This notebook sets up the project, starts an Antioch Isaac Lab session, opens the packaged robot USD in the simulator viewer. Training comes later. The notebook itself does not need a Jupyter kernel.


## 1. Get the repository

If you have **not cloned** it yet:

```bash
mkdir -p ~/projects
cd ~/projects
git clone --depth 1 --single-branch --branch antioch/humanoid_legs_5 https://github.com/triton-droids/simulation.git
cd simulation
```

If you **already have** the repository, open a terminal anywhere inside it:

```bash
cd "$(git rev-parse --show-toplevel)"
git status --short
git fetch origin
if git show-ref --verify --quiet refs/heads/antioch/humanoid_legs_5; then
  git switch antioch/humanoid_legs_5
else
  git switch --track origin/antioch/humanoid_legs_5
fi
git pull --ff-only
```

If Git reports changes that block the switch, save those changes before continuing. Run every remaining block from the `simulation/` root, where `antioch.yaml` and `robots/` are visible.


## 2. Set up Python and Antioch

```bash
uv sync --locked
uv run antioch --version
uv run antioch auth login
```

The terminal may show `(base)` because Conda is installed. That is okay: `uv` runs this project in its own `.venv`, and you do not need to create another Conda environment. If `uv` is missing, install it using the [uv guide](https://docs.astral.sh/uv/getting-started/installation/). Follow Antioch's browser sign-in prompt. This project already has `pyproject.toml` and `antioch.yaml`; do not run `uv init` or `antioch init` again.


## 3. Check the robot bundle

```bash
ls robots/humanoid_legs_5/usd_import/humanoid_legs_5/humanoid_legs_5.usda
find robots/humanoid_legs_5/meshes -name '*.stl' | wc -l
```

The second command should print **24**. The USD is the ready-to-open Isaac Sim asset. The MJCF is kept as its source; onboarding does not require converting it.


## 4. Start an Isaac Lab session on Antioch

```bash
uv run antioch session new
uv run antioch service exec -- python -c 'import isaaclab, isaacsim; print("Isaac Lab:", isaaclab.__file__); print("Isaac Sim:", isaacsim.__file__)'
```

The first command starts a remote GPU session and may take a few minutes. The second confirms both simulator packages are available there. Antioch syncs the local project to `/workspace/project` before remote commands run.


## 5. Open the legs in the simulator window

Run this in your **local terminal**. Antioch opens the imported legs in a streamed Isaac Sim viewport for five minutes:

```bash
uv run antioch service exec --stream -- python src/view_legs.py --seconds 300
```

Open the stream in Antioch Mission Control if it does not appear automatically. Use a larger `--seconds` value if you need more time. This is a visual inspection of the packaged USD; it does not validate ankle physics.


## 6. Finish the session

When you are done viewing the robot, release the GPU session:

```bash
uv run antioch session release
```
